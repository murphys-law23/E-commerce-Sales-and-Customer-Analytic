# E-commerce exploratory analysis
Reproducible exploratory analysis using the provided synthetic dataset.


```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
df = pd.read_csv(Path('../data/orders.csv'), parse_dates=['order_date'])
assert df.order_id.is_unique
assert df.isna().sum().sum() == 0
assert df.discount_pct.between(0,1).all()
completed = df.loc[df.status.eq('Completed')].copy()
completed['net_sales'] = completed.quantity * completed.unit_price * (1-completed.discount_pct)
monthly = completed.groupby(completed.order_date.dt.to_period('M'))['net_sales'].sum()
category = completed.groupby('category')['net_sales'].sum().sort_values(ascending=False)
customer_orders = completed.groupby('customer_id').order_id.nunique()
print('Net sales:',round(completed.net_sales.sum(),2))
print('Average order value:',round(completed.net_sales.sum()/len(completed),2))
print('Repeat customer rate:',round(customer_orders.ge(2).mean(),3))
print(category)
monthly.plot(kind='bar',figsize=(11,4),title='Monthly net sales')
plt.tight_layout();plt.show()
```
